The following script utilizes pandas and numpy to generate 5,000 realistic records. It intentionally bakes in a logical relationship where cooperative membership, high mobile money inflows, and subsidy access reduce the likelihood of default.

```python
import pandas as pd
import numpy as np
```

```python
# Set seed for reproducibility
np.random.seed(42)

n_samples = 5000

# Generate independent variables
farmer_ids = [f"NMB-AGRI-{i:05d}" for i in range(n_samples)]
regions = np.random.choice(["Dodoma", "Arusha", "Mbeya", "Morogoro", "Mwanza"], n_samples)
coop_members = np.random.choice([0, 1], n_samples, p=[0.4, 0.6])
crops = np.random.choice(["Maize", "Cashew", "Coffee", "Rice"], n_samples)
farm_sizes = np.round(np.random.uniform(0.5, 10.0, n_samples), 1)

# Generate alternative financial data
# Base inflow on farm size and crop type multipliers
mobile_money_inflows = farm_sizes * np.random.uniform(500000, 1500000, n_samples)
subsidy_status = np.random.choice([0, 1], n_samples, p=[0.7, 0.3])
loan_amounts = np.round(np.random.uniform(200000, 5000000, n_samples), -3)

# Calculate synthetic default probability based on risk factors
# Higher inflow, coop membership, and subsidies lower the risk
base_risk = 0.4
risk_scores = (
    base_risk 
    - (coop_members * 0.15) 
    - (subsidy_status * 0.1) 
    - (np.log1p(mobile_money_inflows) / 100) 
    + (loan_amounts / 20000000) 
)

# Convert continuous risk score to binary default flag (1 or 0)
probabilities = 1 / (1 + np.exp(-risk_scores)) # Sigmoid function
default_flags = np.random.binomial(1, probabilities)

# Compile dataset
df_credit = pd.DataFrame({
    "farmer_id": farmer_ids,
    "region": regions,
    "coop_member": coop_members,
    "farm_size_hectares": farm_sizes,
    "primary_crop": crops,
    "mobile_money_inflow": mobile_money_inflows,
    "subsidy_status": subsidy_status,
    "loan_amount_requested": loan_amounts,
    "default_flag": default_flags
})

```

```python
df_credit.to_csv("agri_credit_data.csv", index=False)
print("agri_credit_data.csv generated successfully.")
```
