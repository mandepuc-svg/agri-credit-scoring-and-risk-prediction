```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
import category_encoders as ce
```

```python
# 1. Preprocessing
X = df_credit.drop(columns=["farmer_id", "default_flag"])
y = df_credit["default_flag"]

# Encode categoricals (Target Encoding or One-Hot)
encoder = ce.TargetEncoder(cols=["region", "primary_crop"])
X_encoded = encoder.fit_transform(X, y)

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# 2. Model Training
rf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
rf_model.fit(X_train, y_train)

# 3. Predict Probability of Default (PD)
# Get probability of class 1 (Default)
pd_predictions = rf_model.predict_proba(X_test)[:, 1] 

# 4. Calculate Expected Loss (EL)
LGD = 0.45 # Assuming 45% Loss Given Default (to be swapped with another models output)
EAD = X_test["loan_amount_requested"].values # Exposure at Default (to be swapped with another models output)

expected_losses = pd_predictions * LGD * EAD

```

```python
# Display results for top 5 farmers in test set
results = pd.DataFrame({
    "Loan Amount (EAD)": EAD,
    "Probability of Default (PD)": pd_predictions,
    "Expected Loss (EL)": expected_losses
})
print(results.head())
```
